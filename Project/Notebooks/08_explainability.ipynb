Explainability and Interpretability Analysis

Interpretability was included in this project not as a separate technical exercise, but as part of the central comparison between predictive performance and model transparency. Since the task is a health-related binary classification problem, model quality cannot be assessed only by how accurately diabetes status is predicted. It is also useful to understand, at least to some degree, which variables the models appear to rely on and whether those variables are consistent across different modelling approaches. In this project, interpretability was examined using two different methods: coefficient-based interpretation for Logistic Regression, and feature-importance-based interpretation for the tree-based models, Random Forest and Gradient Boosting.

These methods provide different kinds of information. Logistic Regression offers a relatively direct account of the direction and relative size of associations between predictors and the model’s estimated log-odds of the positive class. By contrast, tree-based feature importance measures indicate how much each variable contributes to prediction within a nonlinear ensemble, but they do not show whether the variable increases or decreases the predicted probability of diabetes. As a result, the interpretability analysis is not based on a single unified notion of explanation. Instead, it involves comparing different forms of model-specific interpretation and identifying where they converge.

Interpreting Logistic Regression Coefficients

Logistic Regression is the most straightforward model to interpret in this study because its coefficients have a clear mathematical meaning. A positive coefficient indicates that, holding the other predictors constant, the variable is associated with higher predicted odds of diabetes. A negative coefficient indicates an association with lower predicted odds. Exponentiating the coefficients gives odds ratios, which are often easier to discuss in prose, although they still need careful interpretation.

The strongest positive coefficient in the model is CholCheck (coefficient = 1.320, odds ratio = 3.745). On the surface, this suggests that the presence of a cholesterol check is associated with substantially higher predicted odds of diabetes. However, this is a good example of why interpretability must be handled carefully. It would be inappropriate to read this as evidence that receiving a cholesterol check increases diabetes risk. A more plausible interpretation is that CholCheck may be acting as a proxy for healthcare contact, monitoring intensity, age, or pre-existing health concerns. In other words, the variable is informative for prediction, but the reason it is informative may be indirect. This is precisely the sort of distinction that interpretability analysis should preserve.

HighBP (0.715, odds ratio = 2.045) and HighChol (0.566, odds ratio = 1.762) are also strong positive predictors. These are easier to interpret in a predictive sense, as both variables are plausibly associated with broader metabolic or cardiovascular risk profiles that overlap with diabetes status. GenHlth (0.537, odds ratio = 1.711) is similarly important. Since this is an ordinal variable, the coefficient should be understood as the effect of moving one category upward on the general health scale, rather than a simple binary change. Its prominence suggests that self-reported worsening health status carries substantial predictive information in this dataset.

BMI (0.469, odds ratio = 1.599) is also among the strongest positive coefficients, which is consistent with domain expectations. However, because BMI was standardised before modelling, its coefficient does not correspond to a one-unit raw increase in BMI. Rather, it reflects approximately a one-standard-deviation increase, which means it should not be compared too literally with the binary variables. Even so, its magnitude indicates that BMI is an important contributor to the linear model.

Other positive coefficients, such as Sex, HeartDiseaseorAttack, Stroke, and Age, suggest that the model captures a broader pattern involving demographic and health-related risk markers. Their coefficients are smaller, but still meaningful as part of the overall predictive structure.

The most notable negative coefficient is HvyAlcoholConsump (-0.788, odds ratio = 0.455). This implies that, within the fitted model, this variable is associated with lower predicted odds of diabetes. As with CholCheck, this should not be interpreted causally or too literally. It may reflect confounding, subgroup effects, limitations in how the survey variable is coded, or interactions with other health and lifestyle variables. The key point is that the model found this variable useful in separating cases, not that it reveals a simple protective relationship.

Overall, the Logistic Regression results provide directional interpretability. They make it possible to say which variables are associated with higher or lower predicted diabetes odds, and roughly how strong those associations are within the fitted model. At the same time, they do not justify causal claims, and the meaning of the coefficients depends on variable type, scaling, and the presence of the other predictors in the model.

Interpreting Tree-Based Feature Importances

The interpretability outputs for Random Forest and Gradient Boosting are different in kind. Feature importance in these models reflects how useful a variable was in reducing impurity or improving predictive splits across the ensemble. This means the importance values show relative contribution to prediction, but not direction. A highly important feature is one the model relied on heavily, but the importance score alone does not show whether larger values of that feature push predictions toward or away from the diabetes class.

The Random Forest model distributes importance across a fairly broad set of predictors. Its most important variables are BMI (0.1576), GenHlth (0.1511), HighBP (0.1259), and Age (0.1255), followed by Income, HighChol, PhysHlth, MentHlth, Education, and DiffWalk. This pattern suggests that the forest is using a relatively wide range of information, combining metabolic, general health, physical functioning, and socioeconomic signals. In interpretive terms, this indicates that the model does not rely on a single dominant variable, but instead spreads predictive weight across multiple related dimensions.

The Gradient Boosting model is more concentrated. Its top variables are HighBP (0.3379), GenHlth (0.2541), BMI (0.1324), Age (0.1162), and HighChol (0.0721), after which the importance values drop sharply. Compared with Random Forest, this suggests that Gradient Boosting is placing much greater emphasis on a smaller core set of predictors. This is analytically useful because it gives a clearer account of what the boosting model appears to consider most informative. In this case, the model’s predictive logic seems to be dominated by blood pressure, self-reported general health, BMI, age, and cholesterol status.

Feature importance is helpful because it reveals which variables matter most inside complex nonlinear models that are otherwise harder to inspect directly. However, it also has clear limitations. It does not indicate the sign of an effect, it can be influenced by correlations between predictors, and it is less transparent than a linear coefficient. For example, if two variables contain overlapping information, their importance may be divided unevenly across models. For this reason, tree-based importance should be interpreted as a measure of predictive usefulness within the model, not as a statement about independent causal influence.

Comparing Interpretability Across Model Types

The interpretability methods used here answer related but distinct questions. Logistic Regression addresses a question of direction and relative association: which variables are linked to higher or lower predicted diabetes odds, once the model has fitted a linear boundary? The tree-based models address a question of reliance: which variables were most useful to the nonlinear ensemble when making prediction splits?

This difference matters. A coefficient-based interpretation is more transparent in form, but it depends on the assumption that the model expresses relationships in a linear way on the log-odds scale. A feature-importance-based interpretation is more flexible with respect to nonlinear relationships and interactions, but less precise in explaining how a variable influences the prediction. In other words, Logistic Regression is easier to explain, but it may simplify the structure of the data; Random Forest and Gradient Boosting may capture richer relationships, but their explanations are less direct.

This means the interpretability section should not present all three models as offering the same kind of insight. Rather, the value of comparing them lies in looking for convergence. If different models, using different representational structures and different interpretive tools, repeatedly identify the same predictors as important, confidence increases that those variables are genuinely central to the predictive task.

Stable Predictors Across Models

The strongest result from the cross-model comparison is the consistency of a small set of predictors across the tree-based models, and to a considerable extent across Logistic Regression as well. The shared stable predictors are:

BMI
GenHlth
HighBP
Age
HighChol

These variables form the clearest common signal in the project. Random Forest ranks BMI, GenHlth, HighBP, and Age as its top four variables. Gradient Boosting ranks HighBP, GenHlth, BMI, and Age as its top four, with HighChol also in the top five. Logistic Regression likewise assigns substantial positive coefficients to HighBP, HighChol, GenHlth, BMI, and Age.

This agreement is important because it suggests that the models are not producing entirely model-specific stories. Despite the differences between linear and nonlinear methods, and despite the different interpretation tools used, the same core predictors emerge repeatedly. That strengthens the credibility of the interpretability analysis. It suggests that the predictive structure of the dataset is relatively stable around a cluster of well-known health-related indicators, rather than being driven by arbitrary or highly unstable variable rankings.

At the same time, there are also differences between the models. Random Forest places relatively more emphasis on Income, PhysHlth, MentHlth, and Education, implying a more distributed view of the feature space. Gradient Boosting gives much more dominance to the top few variables, especially HighBP and GenHlth. These differences are themselves informative. They suggest that while there is agreement on the core predictors, the models vary in how broadly they treat secondary variables as contributing to the classification problem.

Variables Requiring Especially Careful Interpretation

Two variables in particular deserve caution: CholCheck and HvyAlcoholConsump.

CholCheck appears as the strongest Logistic Regression coefficient and also enters the Gradient Boosting top ten. This indicates that it is genuinely informative for prediction. However, it is not straightforwardly interpretable as a direct risk indicator in the same way as BMI or high blood pressure. More likely, it reflects an indirect relationship: people who receive cholesterol checks may be older, more engaged with healthcare services, already identified as at-risk, or subject to more clinical monitoring. In that sense, the variable may function as a proxy for healthcare contact or pre-existing concern rather than a direct marker of biological diabetes risk. This does not make it unimportant, but it does mean the interpretation has to remain cautious and contextual.

HvyAlcoholConsump is also difficult to interpret cleanly. Its negative Logistic Regression coefficient and small but non-negligible Gradient Boosting importance show that the variable contributes to prediction, but not in a way that supports a simple substantive claim. The negative sign in the linear model should not be taken to mean that heavy alcohol consumption reduces diabetes risk. More plausibly, the variable may be capturing a more complex pattern in the dataset, potentially shaped by confounding, differences in subgroup composition, or limitations in how the survey measure is defined. This is a good example of why predictive importance and substantive explanation are not the same thing.

These cases illustrate a broader methodological point: interpretability outputs are not self-interpreting. Some variables align easily with domain expectations, while others require restraint because they may be acting as proxies, composites, or artefacts of measurement. A careful report should acknowledge both kinds.

Contribution of the Interpretability Analysis

The interpretability analysis adds an important second layer to the project. The evaluation metrics indicate how well the models perform, but they do not show which features are driving those predictions. By examining coefficients and feature importances, it becomes possible to assess whether the models appear to rely on plausible and stable predictors, and whether the explanatory picture changes across model families.

In this project, the explainability results show substantial convergence around a core group of predictors: BMI, general health, high blood pressure, age, and high cholesterol. This gives the modelling results more substantive coherence, since the strongest predictive signals are broadly consistent across both linear and tree-based approaches. At the same time, the analysis also highlights the limits of interpretability. Some variables, particularly CholCheck and HvyAlcoholConsump, may be important for prediction without having a simple direct interpretation. This reinforces the need to distinguish predictive usefulness from causal explanation.

Taken together, the explainability section strengthens the overall project by showing not only which models performed well, but also what kinds of information those models appear to be using. That is especially relevant in a health-related context, where model transparency and the careful interpretation of predictive signals are important alongside raw performance.
