Full Baseline Model Comparison

To extend the initial baseline analysis, two additional ensemble methods were trained and evaluated: Random Forest and Gradient Boosting. These were compared against the earlier Logistic Regression and Decision Tree models using the same held-out test set and the same evaluation metrics.

Interpretation

The four-model comparison shows a clear performance hierarchy. Decision Tree was the weakest model, producing the lowest values across all reported metrics. This suggests that a single tree is not sufficient for this diabetes prediction task and that more robust methods are needed to capture the relevant relationships in the data.

Logistic Regression remained a strong and competitive baseline. Although it was slightly outperformed by the ensemble methods, it achieved solid recall and ROC-AUC while retaining the advantage of high interpretability. This is important because the project is concerned not only with predictive performance, but also with model transparency and practical interpretability.

Among all models, Random Forest provided the best overall balance of performance. It achieved the highest accuracy, precision, and F1-score, while also maintaining strong recall and ROC-AUC. This suggests that Random Forest is currently the most balanced model when considering overall classification performance.

Gradient Boosting achieved the highest recall and ROC-AUC, meaning that it was the strongest model for identifying positive diabetes cases and for separating the two classes across thresholds. This is particularly relevant in a health-related classification setting, where false negatives are often more serious than false positives. However, the increase in recall came with a slightly higher number of false positives.

Overall, the results suggest that the ensemble methods improve upon both Logistic Regression and Decision Tree, with Random Forest emerging as the strongest all-round performer and Gradient Boosting as the best model for maximising sensitivity to positive cases. At the same time, the relatively competitive performance of Logistic Regression means that interpretability remains a meaningful part of the performance-interpretability trade-off in this study.

These results also show that accuracy alone would be misleading in this imbalanced classification problem. The more informative metrics here are recall, F1-score, and ROC-AUC, since they better reflect the models’ ability to detect diabetes-positive cases while accounting for the trade-off between missed cases and false alarms.

At this stage, Random Forest and Gradient Boosting appear to be the strongest candidates for further tuning, while Logistic Regression remains the most useful interpretable benchmark.
