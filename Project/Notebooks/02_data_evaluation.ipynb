## Data Quality Notes

- No missing values were found in any column, so missing-data handling is unlikely to be a major issue.
- The main data-quality concern is the presence of **24,206 duplicate rows** (**9.54%** of the dataset), which will need to be considered during preprocessing.
- All binary variables appear to be coded consistently as **0/1**, with no obvious invalid codes detected.
- The ordinal/coded variables also have sensible ranges:
  - `Age`: 1-13
  - `GenHlth`: 1-5
  - `Education`: 1-6
  - `Income`: 1-8
- `BMI` ranges from **12 to 98**, which suggests some extreme values are present. Using a simple threshold check, **873 rows** have `BMI < 15` or `BMI > 60`, so these potential outliers should be reviewed later.
- `MentHlth` and `PhysHlth` range from **0 to 30**, which is plausible for count-style health-day variables.
- Overall, the dataset appears well coded and relatively clean, so preprocessing should be fairly straightforward apart from decisions about duplicate rows and possible extreme BMI values.
