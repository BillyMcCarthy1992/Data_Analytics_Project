## Quick Dataset Notes

- **Dataset size:** 253,680 rows, 22 columns  
- **Target variable:** `Diabetes_binary`  
- **Target values:** 0 and 1, so the task is already **binary classification**  
- **Class balance:** 86.1% class 0, 13.9% class 1 → **moderate imbalance**  
- **Missing values:** none  
- **Duplicate rows:** 24,206 → important preprocessing issue  
- **Most predictors** are binary indicators  
- **`GenHlth`, `Age`, `Education`, and `Income`** appear ordinal  
- **`BMI`** appears continuous; **`MentHlth`** and **`PhysHlth`** appear count/discrete  
- The dataset looks well aligned with the planned **diabetes prediction project**  
- It is a good candidate for comparison using **Logistic Regression, Decision Tree, Random Forest, and Gradient Boosting/XGBoost**
