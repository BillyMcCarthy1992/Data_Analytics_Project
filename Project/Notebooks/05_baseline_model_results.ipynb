Baseline Modelling: Logistic Regression and Decision Tree

After completing the data cleaning and preprocessing steps, two baseline classification models were trained and evaluated on the held-out test set: Logistic Regression and Decision Tree. The purpose of this stage was to establish an initial benchmark for diabetes prediction using the selected health indicators before moving on to more advanced ensemble methods.

The dataset had already been cleaned by removing exact duplicate rows, and the data was split using an 80/20 stratified train-test split to preserve the class distribution of the target variable, Diabetes_binary. For Logistic Regression, the numeric variables (BMI, MentHlth, and PhysHlth) were standardised, while binary and ordinal variables were left unchanged. The Decision Tree model used the unscaled predictors directly, as tree-based methods do not require feature scaling.

Both models were trained with class_weight="balanced" in order to account for the class imbalance in the dataset, where the diabetes-positive class is the minority class. Model performance was evaluated using Accuracy, Precision, Recall, F1-score, and ROC-AUC, along with confusion matrices and classification reports.

Interpretation

The results show that Logistic Regression outperformed the Decision Tree across all evaluation metrics. This is especially important because the project is concerned not only with overall predictive performance, but also with identifying positive diabetes cases reliably.

The most important metric in this context is recall, since false negatives are particularly undesirable in a health-related prediction task. Logistic Regression achieved a recall of 0.7598, compared with 0.7377 for the Decision Tree. This means that Logistic Regression correctly identified a slightly greater proportion of true diabetes-positive cases.

This can also be seen in the confusion matrices:

Logistic Regression correctly identified 5,333 positive cases and missed 1,686
Decision Tree correctly identified 5,178 positive cases and missed 1,841

Therefore, Logistic Regression missed 155 fewer positive cases than the Decision Tree on the test set.

Although Logistic Regression performed better overall, both models showed low precision, indicating that a substantial number of predicted positive cases were actually false positives. For example, Logistic Regression produced 11,434 false positives, while the Decision Tree produced 11,782 false positives. This suggests that both models prioritise sensitivity over precision, which is a common trade-off when handling imbalanced health data with balanced class weighting.

Another important observation is that accuracy is not the most informative metric in this task. Because the dataset is imbalanced, a naive model predicting only the majority class could achieve a high accuracy while completely failing to detect diabetes-positive cases. For that reason, recall, F1-score, and ROC-AUC provide a more meaningful basis for comparison here than accuracy alone.

Overall, these baseline results suggest that Logistic Regression provides the stronger initial balance between predictive performance and interpretability. It not only achieved the best recall, F1-score, and ROC-AUC, but is also more interpretable than a tree model, making it a strong benchmark for the next stage of the project. The next step is to extend the comparison by training more advanced models, such as Random Forest and Gradient Boosting/XGBoost, to determine whether increased model complexity leads to a meaningful improvement in performance.
