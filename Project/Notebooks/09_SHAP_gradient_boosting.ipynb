SHAP Analysis of the Tuned Gradient Boosting Model

To get a more detailed explanation of the tuned Gradient Boosting model, I applied SHAP to a sample of the data. Unlike the model’s built-in feature importance scores, SHAP gives a clearer picture of how much each feature contributes to the model output and, in the beeswarm plot, also shows whether higher or lower values tend to push predictions toward or away from the diabetes class.

The SHAP bar plot ranked features by their mean absolute SHAP value, so it shows which variables had the largest overall influence on predictions across the explained sample. The most important features were:

GenHlth
Age
BMI
HighBP
HighChol

These were followed by Sex, Income, HvyAlcoholConsump, MentHlth, and CholCheck.

This result suggests that the tuned Gradient Boosting model relies mainly on a fairly stable core group of health-related predictors. In particular, general health, age, BMI, high blood pressure, and high cholesterol appear to be the strongest overall drivers of the model’s predictions. This is useful because it broadly matches the earlier feature-importance analysis, which increases confidence that the model is not relying on random or highly unstable signals.

The SHAP beeswarm plot adds more detail than the bar plot. It shows both the size of each feature’s contribution and the direction of that contribution. Points further to the right push the model output more toward the diabetes class, while points to the left push it away. The colour also matters: red/pink points represent higher feature values, while blue points represent lower feature values.

The beeswarm plot showed a clear pattern for the most important variables:

For GenHlth, higher values tended to push predictions toward diabetes, while lower values pushed them away. Since higher GenHlth values correspond to poorer general health, this is a sensible result.
For Age, higher values were generally associated with positive SHAP values, meaning older age tended to increase the model’s diabetes prediction.
For BMI, higher values also tended to push the prediction toward diabetes, while lower values tended to reduce it.
For HighBP and HighChol, the two categories separated quite clearly, with the higher-risk category contributing positively to diabetes predictions.

This gives a much more interpretable picture of the tuned Gradient Boosting model. It is not just that these features are important in a general sense; the SHAP results show that higher values of key health-risk variables tend to increase the model output toward the positive diabetes class.

Some variables still need more cautious interpretation. HvyAlcoholConsump and CholCheck both appeared among the more important SHAP features, which means the model is genuinely using them, but their meaning is less straightforward. For example, CholCheck may be acting as a proxy for healthcare contact or prior monitoring rather than representing a direct risk signal on its own. Similarly, HvyAlcoholConsump should not be interpreted causally just because it contributes to prediction in the model. In both cases, the SHAP results show predictive importance, not proof of a simple real-world causal relationship.

Overall, the SHAP analysis strengthens the interpretability section by showing that the tuned Gradient Boosting model is driven mainly by a coherent set of predictors, especially general health, age, BMI, high blood pressure, and high cholesterol. It also provides a more detailed explanation than ordinary feature importance by showing both which variables matter most and how they tend to influence the model’s predictions.
